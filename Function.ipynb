```python
#calculate final price
def calculate_discount(price,discount_percentage):
  discount_amount = price * (discount_percentage/100)
  final_price = price - discount_amount
  return final_price
```

```python
calculate_discount(1000,20)
```

```python
#calculate GST
def calculate_gst(price,gst_percentage):
  gst_amount = price * (gst_percentage/100)
  final_price = price + gst_amount
  return final_price
```

```python
calculate_gst(1000,5)
```

```python
#calculate delivery charges
def calculate_delivery_charges(price,delivery_charges):
  delivery_charges_amount = price * (delivery_charges/100)
  final_price = price + delivery_charges_amount
  return final_price
```

```python
delivery_charge = calculate_delivery_charges(1000,40)
print(delivery_charge)
```
